```python
"""
8. Write a program for n gram language modelling from scratch using numpy and other 
relevant modules to predict next word, compare results of unigram, bigram, trigram using 
perplexity. Store all the n-gram probability in dictionary and then use it for inference. 
"""
```

```python
from collections import defaultdict

def create_ngrams(corpus, n):
    ngrams = []

    for doc in corpus:
        document = ['<s>'] * (n-1) + doc.split(' ') + ['</s>'] * (n-1)
        for index in range(0, len(document) - n + 1):
            ngrams.append(tuple( doc.lower() for doc in document[index:index+n]))

    return ngrams

def calculate_word_probability(corpus):

    words = []
    for document in corpus:
        words.extend(document.split(' '))

    total_words = len(words)
    probabilities = defaultdict()
    for word in words:
        probabilities[word.lower()] = words.count(word) / total_words
            
    return probabilities

def calculate_probabilities(ngrams):
    probabilities = defaultdict()
    total_ngrams = len(ngrams)

    keys = set(ngrams)
    for ngram in keys:
        probabilities[ngram] = ngrams.count(ngram) / total_ngrams
            
    return probabilities
```

```python
corpus = """
I like machine learning.
I like deep learning.
I like natural language processing.
I study machine learning.
I study deep learning.
I study natural language processing.
Machine learning is interesting.
Machine learning is useful.
Deep learning is interesting.
Deep learning is powerful.
Natural language processing is interesting.
Natural language processing is useful.
I enjoy learning machine learning.
I enjoy learning deep learning.
I enjoy natural language processing.
Students study machine learning.
Students study deep learning.
Students study natural language processing.
Machine learning helps students.
Deep learning helps students.
""".split('\n')

test_sentences = [
    "I like",
    "I study",
    "machine",
    "machine learning",
    "deep",
    "deep learning",
    "natural language",
    "students study",
    "I enjoy learning"
]

corpus = corpus[1:-1]
# corpus, test_sentences
```

```python
unigrams = create_ngrams(corpus, 1)
bigrams = create_ngrams(corpus, 2)
trigrams = create_ngrams(corpus, 3)
```

```python
unigram_prob = calculate_probabilities(unigrams)
bigram_prob = calculate_probabilities(bigrams)
trigram_prob = calculate_probabilities(trigrams)
word_prob = calculate_word_probability(corpus)
```

```python
test_statement = "I enjoy learning".lower()
test_tokens = test_statement.split(' ')

test_tokens
```

```python
def calculate_nextword_probability(ngram_prob, word_prob, n, test_tokens):
    probability = 1
    for i in range(len(test_tokens) - n + 1):
        probability = probability * (ngram_prob[tuple(test_tokens[i:i+n])] / word_prob[test_tokens[i]] )
        print( tuple(test_tokens[i:i+n]), ngram_prob[tuple(test_tokens[i:i+n])], word_prob[test_tokens[i]] )

    return probability

ugp = calculate_nextword_probability(unigram_prob, word_prob, 1, test_tokens)
bgp = calculate_nextword_probability(bigram_prob, word_prob, 2, test_tokens)
tgp = calculate_nextword_probability(trigram_prob, word_prob, 3, test_tokens)
```
