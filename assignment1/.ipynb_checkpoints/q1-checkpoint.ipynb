### Preprocessing Pipeline

```python
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

def lower(text):
    return text.lower()

def remove_punct(text):
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'https?://\S+', '', text)
    return text

def tokenize(text):
    return word_tokenize(text)

def stop_words(tokens):
    stop_words = stopwords.words('english')
    tokens = [token for token in tokens if token not in stop_words and token.isalpha()]
    return tokens 

def stemming(tokens):
    stemmer = PorterStemmer()
    return [ stemmer.stem(token) for token in tokens ]

def lemmatizing(tokens):
    lemmatizer = WordNetLemmatizer()
    return [ lemmatizer.lemmatize(token) for token in tokens ]


def preprocessing(text):
    text = lower(text)
    text = remove_punct(text)
    tokens = word_tokenize(text)
    tokens = stop_words(tokens)
    tokens = stemming(tokens)
    tokens = lemmatizing(tokens)
    return ' '.join(tokens)

text = "This is sample text containing    impurity like https://abc.net  all of that punctuations,."
preprocessing(text)
```

### Text file - sample.txt

```python
with open("sample.txt", "r") as ptr:
    data = ptr.read()
    txt_df = pd.DataFrame(data.split('\n'), columns=["text"] )
txt_df
```

```python
txt_df['text'] = txt_df['text'].apply(preprocessing)
txt_df
```

### CSV - sample.csv

```python
csv_df = pd.read_csv("sample.csv")
csv_df
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
csv_df['Topic'] = encoder.fit_transform(csv_df['Topic'])

csv_df['Comment'] = csv_df['Comment'].apply(preprocessing)
csv_df
```

### Library - movie_reviews

```python
from nltk.corpus import movie_reviews

lib_data = np.array([ (movie_reviews.raw(fileid), movie_reviews.categories(fileid)[0] ) for fileid in movie_reviews.fileids() ])
lib_df = pd.DataFrame(lib_data, columns = ["text", "label"])
lib_df.head()
```

```python
lib_df["text"] = lib_df["text"].apply(preprocessing)
lib_df.head()
```
