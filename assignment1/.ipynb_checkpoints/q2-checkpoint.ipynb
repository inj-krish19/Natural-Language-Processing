```python
"""
2. Implement a lemmatizer using the WordNet and compare its output with stemming. Use 
the Gutenberg Corpus.   
"""
```

```python
import numpy as np
import pandas as pd 
import re, nltk
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

def lower(text):
    return text.lower()

def remove_punct(text):
    text = re.sub(r'[^\w\s]', '', text)
    text = re.sub(r'https?://\S+', '', text)
    return text

def tokenize(text):
    return word_tokenize(text)

def stop_words(tokens):
    stop_words = stopwords.words('english')
    tokens = [token for token in tokens if token not in stop_words and token.isalpha()]
    return tokens 

def stemming(tokens):
    stemmer = PorterStemmer()
    return [ stemmer.stem(token) for token in tokens ]

def lemmatizing(tokens):
    lemmatizer = WordNetLemmatizer()
    return [ lemmatizer.lemmatize(token) for token in tokens ]


def preprocessing_stemming(text):
    text = lower(text)
    text = remove_punct(text)
    tokens = word_tokenize(text)
    tokens = stop_words(tokens)
    tokens = stemming(tokens)
    return ' '.join(tokens)

def preprocessing_lemmatizing(text):
    text = lower(text)
    text = remove_punct(text)
    tokens = word_tokenize(text)
    tokens = stop_words(tokens)
    tokens = lemmatizing(tokens)
    return ' '.join(tokens)
```

```python
nltk.download('gutenberg')
from nltk.corpus import gutenberg

data = [ gutenberg.raw(fileid) for fileid in gutenberg.fileids() ]
df = pd.DataFrame(data, columns = ["text"])
df
```

### Lemmatizing

```python
# Lemmatizing
df['text'] = df['text'].apply(preprocessing_lemmatizing)
df
```

### Stemming 

```python
# Stemming
df['text'] = df['text'].apply(preprocessing_stemming)
df
```
