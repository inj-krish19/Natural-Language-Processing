```python
"""
7. Write a program to find edit distance between any two string, output should be in pandas 
data frame 
"""
```

```python
import numpy as np
import pandas as pd
```

```python
source = "ELEPHANT"
destination = "RELEVANT"

source_length = len(source)
destination_length = len(destination)

source, destination, source_length, destination_length
```

```python
table = np.zeros((destination_length+1, source_length+1), dtype=np.uint8)

for index in range(1, destination_length+1):
    table[index, 0] = index

for index in range(1, source_length+1):
    table[0, index] = index

table
```

```python
for row in range(1, destination_length+1):
    for col in range(1, source_length+1):
        if source[col-1] == destination[row-1]:
            table[row, col] = table[row-1, col-1]
        else:
            table[row, col] = 1 + min(table[row-1, col-1], table[row-1, col], table[row, col-1])
```

```python
table_df = pd.DataFrame(table, columns = ['#'] + list(source), index = ['#'] + list(destination) )
table_df
```

```python
print(f"Edit Distance of {source} to {destination} is {table[-1, -1]}")
```
