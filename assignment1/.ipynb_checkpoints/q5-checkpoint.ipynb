```python
"""
5. Implement TF-IDF vectorization and apply it to a text corpus to extract important 
features. Use the IMDB movie reviews dataset.   
"""
```

```python
import nltk
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
from nltk.corpus import movie_reviews

data = [ (movie_reviews.raw(fileid), movie_reviews.categories(fileid)[0]) for fileid in movie_reviews.fileids() ]
df = pd.DataFrame( data[0:100] + data[-100:], columns = ["text", "category"])
df
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
vectors = tfidf.fit_transform(df['text']).toarray()
features = tfidf.get_feature_names_out()

average_scores = np.mean(vectors, axis=0)
top_indices = np.argsort(average_scores)[-20:][::-1]

print("Top 20 Important Features:")
for i in top_indices:
    print(features[i], average_scores[i])
```
