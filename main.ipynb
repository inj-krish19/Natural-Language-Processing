```python
def lower(doc):
    return doc.lower()

def punc(doc):
    new = re.sub(r'[^\w\s]', "", doc)
    new = re.sub(r'https?://s+', "", doc)
    return new

def token(doc):
    return word_tokenize(doc)

def stop(doc):
    sw = set(stopwords.words('english'))
    rs = [ i for i in doc if i not in sw ]
    return rs

def stem(doc):
    st = PorterStemmer()
    gs = [ st.stem(i) for i in doc]
    return gs
```

```python
def preprocess_text(doc):
    l = lower(doc)
    p = punc(l)
    t = token(p)
    stp = stop(t)
    stm = stem(stp)
    return " ".join(stm)
```

```python
from nltk.corpus import movie_reviews
data = [
    (movie_reviews.raw(fileid), category) 
    for category in movie_reviews.categories() 
    for fileid in movie_reviews.fileids(category)
]

documents, labels = [data[i][0] for i in range(len(data))], [data[i][1] for i in range(len(data))]
df=pd.DataFrame()
df["review"]=documents
df["label"]=labels
df
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['labels'] = le.fit_transform(df['label'])
df
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

vectorizer = TfidfVectorizer()
x = vectorizer.fit_transform(df['review'])
print(vectorizer.get_feature_names_out())
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, df.labels, test_size=0.2)

LGE = LogisticRegression()
model = LGE.fit(X_train,y_train)
```

```python
prediction = model.predict(X_test)
prediction
```

```python
report = classification_report(prediction,y_test)
print(report)
```
